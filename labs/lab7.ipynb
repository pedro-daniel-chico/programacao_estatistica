```python
import pandas as pd
import matplotlib.pyplot as plt 
import numpy as np
```

```python
def funcao(x):
    f = 20*x*(1-x)**3
    return f
```

```python
n=100_000
amostra = []
sucesso = 0
tentativas = 0
while sucesso < n:
    
    u = np.random.uniform(0,1)
    c = 2.2
    y = np.random.uniform(0,1)
    if  u < funcao(y)/c:
        amostra.append(y)
        sucesso+=1
        
    tentativas +=1
x=np.linspace(0,1,10_000)
plt.hist(amostra,density=True)
plt.plot(x,funcao(x))

    
```

```python
np.max(funcao(x))
```

```python
tentativas/sucesso
```

Probabilidade teórica de aceitação do método é 1/c

```python
prob_teorica = 1/c
print(prob_teorica)
```

O caso que temos aqui é quantidade de tentativas até o primeiro sucesso, isso seria uma geométrica, então como a prob de aceitação
é 1/c a média vai ser 1/1/c = c. Ou seja, a média teórica seria c = 2.2 (no meu caso aqui), e a média empírica que calculei tentativas / sucesso = 2.19
