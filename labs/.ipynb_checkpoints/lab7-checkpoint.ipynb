```python
import pandas as pd
import matplotlib.pyplot as plt 
import numpy as np
```

```python
def funcao(x):
    f = 20*x*(1-x)**3
    return f
```

```python
n=100_000
amostra = []
sucesso = 0
while sucesso < n:
    u = np.random.uniform(0,1)
    c = 2.2
    y = np.random.uniform(0,1)
    if  u < funcao(y)/c:
        amostra.append(y)
        sucesso+=1
x=np.linspace(0,1,100)
plt.hist(amostra,density=True)
plt.plot(x,funcao(x))

    
```
