```python
'''
Aqui eu defini uma amostra de uma uniforme, depois criei uma função que recebe a lista que vai ter a amostra e um valor que seria o parâmetro p.
Se o valor for menor que o parâmetro p ele retorna 1 e se for maior retorna 0, a função é de uma bernoulli.
Na função vou colocar para calcular o valor esperado e a variância também
'''
amostra = np.random.uniform(0,1,10000)
def bernoulli(amostra,p):
  amostra_final = []
  valor_esperado = p
  var = p*(1-p)
  print(f"A esperança da bernoulli com parâmetro {p} é {valor_esperado} e a variância é {var}.")
  for i in amostra:
    if i < p:
      amostra_final.append(1)
    else:
      amostra_final.append(0)
  media = np.mean(amostra_final)
  var = np.var(amostra_final)
  print(f"A variância amostral foi de {var} e a média foi de {media}")
  print(amostra_final)




```

```python

#Aqui eu gerei as amostras da multinomial.
m=100
ultima_amostra = []
for j in range(m):
    multi = []
    uniforme=np.random.uniform(0,1,size = m)
    
    for i in range(len(uniforme)):
        
        if 0 < uniforme[i] < 0.10:
            multi.append(1)
        elif 0.10 < uniforme[i] < 0.30:
            multi.append(2)
        elif 0.30<uniforme[i]<0.60:
            multi.append(3)
        else:
            multi.append(4)
    ultima_amostra.append(multi)
#print(ultima_amostra)
```

```python
#Aqui utilizando o np.unique(i,return_counts=True) adicionei os elementos dessa lista na lista multinomial, assim gerei a amostra de multinomial.
multinomial = []
for i in ultima_amostra:
    multinomial.append([i.count(1), i.count(2), i.count(3), i.count(4)])
#print(multinomial)
```
