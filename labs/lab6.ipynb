```python
m = 10000
lam = 4
amostra_poisson = []

for i in range(m):
    soma = 0
    cont = 0
    while soma <= 1:
        u = np.random.uniform(0,1)
        soma += -np.log(1 - u) / lam
        cont += 1
    amostra_poisson.append(cont)
```

```python
fig ,ax = plt.subplots(1,2,figsize=(10,5))
ax[0].hist(amostra_poisson,density = True, bins = 15)
ax[1].hist(np.random.poisson(4,size=m),density = True,bins=15)
plt.show
```
