```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import time as time
```

```python
n=10000
m = 0.5
var = 1/12
amostra = []
amostra_final = []
for i in range(n): #Aqui estou gerando os vetores com as médias.
  u = np.random.uniform(0,1,size=n)
  media_amostral = np.mean(u)
  amostra.append(media_amostral)

for x in amostra: #Aqui eu normalizo os vetores com a média e adiciono na minha amostra final.
  normal = ((n**0.5)*(x-m))/(var)**0.5
  amostra_final.append(normal)
plt.hist(amostra_final)
```

```python
def amostra_bernoulli(n,p):
  amsotra = []
  for u in np.random.uniform(0,1,size = n):
    if p < u:
      amsotra.append(0)
    else:
      amsotra.append(1)
  return amsotra

```

```python
n=1000
p=0.5
var = 1/4
m = 1/2
amostra = []
amostra_final = []
for i in range(n):
  media = np.mean(amostra_bernoulli(n,p))
  amostra.append(media)
for x in amostra:
  normal = ((n**0.5)*(x-m))/(var)**0.5
  amostra_final.append(normal)
plt.hist(amostra_final)
```

```python
def geometrica(p):
  n=1000
  amostra = []
  uniforme = np.random.uniform(0,1,size=n)
  for u in uniforme:
    geo = np.ceil((np.log(1-u))/(np.log(1-p)))
    amostra.append(geo)

  return amostra
```

```python
n=1000
p=0.5
var = 2
m = 2
amostra = []
amostra_final = []

for i in range(n):
  media = np.mean(geometrica(p))
  amostra.append(media)

for x in amostra:
  normal = ((n**0.5)*(x-m))/(var)**0.5
  amostra_final.append(normal)
plt.hist(amostra_final)
```

```python
def poisson(lam):
    n=10000
    poissons = []
    for j in range(n):
        uniforme = np.random.uniform(0,1)
        i = 0
        p = np.exp(-lam)
        F = p
        while uniforme>F:
            i +=1
            p = p*(lam/i)
            F+=p
        poissons.append(i)
    return poissons
    
    
```

```python
n=1000
p=0.5
var = 5
m = 5
amostra = []
amostra_final = []
for i in range(n):
  media = np.mean(poisson(5))
  amostra.append(media)
for x in amostra:
  normal = ((n**0.5)*(x-m))/(var)**0.5
  amostra_final.append(normal)
plt.hist(amostra_final)
```

Para os casos que fiz acima, é possível ver que diminuindo o tamanho de n o histograma não se parece com o de uma normal, ou seja, não aproxima.
Agora aumentando n ele se parece com uma normal.

```python
m=1000
lam = 5
n= 1000
amostra_poisson = []
for i in range(m):
    i=0
    p= lam/n
    pi=(1-p)**n
    F = pi
    U = np.random.uniform(0,1)
    while U > F:
        pi = pi*((n-i)/(i+1)*(p/(1-p)))
        i+=1
        F+=pi
    amostra_poisson.append(i)

fig, ax = plt.subplots(1,3, figsize=(12,10))
ax[0].hist(amostra_poisson)
ax[1].hist(np.random.poisson(5,size=m))
ax[2].hist(poisson(5))


```

Comparando os três métodos pedidos é possível ver que realmente se assemelham.

O tempo de execução entre o método da inversão e o da aproximação pela binomial é diferente, porém bem baixo, o método da aproximação pela binomial é mais rápido que o método recursivo.
Vantagem do método recursivo é que eu não preciso ter uma amostra grande para conseguir simular a distribuição, e a desvantagem do método pela binomial é que preciso de um n grande para conseguir simular.
