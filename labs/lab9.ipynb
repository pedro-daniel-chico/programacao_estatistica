```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
```

Aqui fiz uma função da densidade da gama.

```python
def gamma(x,a,b):
    return (b**a)/(math.gamma(a)) * (x**(a-1)) * np.exp(-b*x)
x = np.linspace(0,3,100)
```

Aqui eu fiz a gama por soma de exponenciais.

```python
lam = 10
n = 10_000
k=5
u = np.random.uniform(0,1,size=(n,k)) #Fiz uma matriz com n linhas e k colunas, todas as entradas sendo valores entre 0,1
exp = -np.log(u)/lam #Aqui eu apliquei a inversa da F(x) da exponencial em cada entrada da matriz, assim eu fiz uma amostra de n por k de todas as entradas exponenciais.
gama = np.sum(exp,axis=1) #Aqui eu somei as colunas, como k = 5, então eu somei cinco exponenciais e gerei então uma amostra de 10_000 gamas.
print(gama)
```

```python
plt.hist(gama,bins=30, density=True)
plt.plot(x,gamma(x,5,10))
```
